## 1. Reversed String

```python
def reverse_string(st):
    if not st:
        print('Empty string cannot be reversed')
    elif len(st) == 1:
        print('single charater cannot be reversed')
    else:
        reversed = ''
        i = -1
        for ch in st:
            reversed = reversed + st[i]
            i = i-1
        return reversed
```

```python
reverse_string('I am cool')
```

## 2. Count Vowels

```python
def vowel_count(st):
    vowels = 'aeiouAEIOU'
    count = 0
    for ch in st:
        for vowel in vowels:
            if ch == vowel:
                count = count+1
                break

    return count
```

```python
vowel_count('Captain America is also cool')
```

```python
def prime_checker(num):
    if num==1:
        return True
    if num<1:
        return False
    for n in range(2, int(num/2)):
        if num%n==0:
            return False

    return True
```

```python
prime_checker(1)
```
